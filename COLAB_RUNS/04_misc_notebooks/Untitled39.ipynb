```python
# Python script to implement Mirzakhani functions D and R based on Lemma 3.1
# and test Equation 3.5

import math

def D_lemma31(x, y, z):
    """Implements Mirzakhani's D function based on Equation 3.1."""
    try:
        numerator = math.exp(x / 2) + math.exp((y + z) / 4)
        denominator = math.exp(-x / 2) + math.exp((y + z) / 4)
        # Ensure denominator is not zero and argument to log is positive
        if denominator <= 0:
            print(f"Warning: Denominator non-positive in D({x}, {y}, {z})")
            return float('nan')
        arg = numerator / denominator
        if arg <= 0:
            print(f"Warning: Argument to log non-positive in D({x}, {y}, {z})")
            return float('nan')
        return 2 * math.log(arg)
    except OverflowError:
        print(f"OverflowError encountered in D({x}, {y}, {z})")
        return float('inf') # Or handle appropriately
    except Exception as e:
        print(f"Error calculating D({x}, {y}, {z}): {e}")
        return float('nan')

def R_lemma31(x, y, z):
    """Implements Mirzakhani's R function based on Equation 3.2."""
    try:
        cosh_y_half = math.cosh(y / 2)
        cosh_xpz_half = math.cosh((x + z) / 2)
        cosh_xmz_half = math.cosh((x - z) / 2)

        numerator = cosh_y_half + cosh_xpz_half
        denominator = cosh_y_half + cosh_xmz_half

        # Ensure denominator is not zero and argument to log is positive
        if denominator <= 0:
            print(f"Warning: Denominator non-positive in R({x}, {y}, {z})")
            return float('nan')
        arg = numerator / denominator
        if arg <= 0:
            print(f"Warning: Argument to log non-positive in R({x}, {y}, {z})")
            return float('nan')
        return x - math.log(arg)
    except OverflowError:
        print(f"OverflowError encountered in R({x}, {y}, {z})")
        return float('inf') # Or handle appropriately
    except Exception as e:
        print(f"Error calculating R({x}, {y}, {z}): {e}")
        return float('nan')

def test_equation_3_5(x1, x2, x3):
    """Tests Mirzakhani's Equation 3.5: R(x1, x2, x3) + R(x1, x3, x2) = x1 + D(x1, x2, x3)."""
    r123 = R_lemma31(x1, x2, x3)
    r132 = R_lemma31(x1, x3, x2)
    d123 = D_lemma31(x1, x2, x3)

    if math.isnan(r123) or math.isnan(r132) or math.isnan(d123):
        print(f"Cannot test Eq 3.5 for ({x1}, {x2}, {x3}) due to NaN result.")
        return float('nan'), float('nan'), float('nan')

    lhs = r123 + r132
    rhs = x1 + d123
    difference = lhs - rhs

    print(f"Testing Eq 3.5 for x1={x1}, x2={x2}, x3={x3}:")
    print(f"  R({x1}, {x2}, {x3}) = {r123}")
    print(f"  R({x1}, {x3}, {x2}) = {r132}")
    print(f"  D({x1}, {x2}, {x3}) = {d123}")
    print(f"  LHS (R123 + R132) = {lhs}")
    print(f"  RHS (x1 + D123) = {rhs}")
    print(f"  Difference (LHS - RHS) = {difference}")
    print("---")
    return lhs, rhs, difference

# --- Test Cases ---
print("Running Test Cases for Mirzakhani Lemma 3.1 Functions and Eq 3.5")

# Test Case 1: Original values where discrepancy was noted
test_equation_3_5(1.0, 2.0, 3.0)

# Test Case 2: Different positive values
test_equation_3_5(2.5, 1.5, 0.5)

# Test Case 3: Values including zero (check edge cases if applicable, though lengths are usually > 0)
# test_equation_3_5(1.0, 0.0, 2.0) # Might need care depending on geometric interpretation

# Test Case 4: Larger values
test_equation_3_5(10.0, 5.0, 15.0)

# Test Case 5: Small values
test_equation_3_5(0.1, 0.2, 0.3)

print("Test Cases Completed.")

```
