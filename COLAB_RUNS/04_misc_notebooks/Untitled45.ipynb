```python
import matplotlib.pyplot as plt
import math

def quark_mass(n):
    # Adjusted formula that outputs mass in MeV directly
    λ = 0.9931
    ε = 0.35
    ω = 3.883
    φ = 7.754687
    log10_K_adjusted = -0.040715  # Combined constant with the factor of 1000 absorbed
    θ = ω * n + φ
    log_mass = λ * n + ε * math.sin(θ) + log10_K_adjusted
    mass_mev = 10 ** log_mass
    return mass_mev

# Define quark names associated with spiral index n = 0 to 5.
quark_names = ['Up', 'Down', 'Strange', 'Charm', 'Bottom', 'Top']
indices = list(range(len(quark_names)))
masses = [quark_mass(n) for n in indices]

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(indices, masses, 'bo-', label='Predicted Masses')
plt.yscale('log')
plt.xlabel('Quark Index (n)')
plt.ylabel('Mass (MeV)')
plt.title('Canonical Star–Ely Quark Spiral Predicted Masses')

# Annotate each point with the quark name and mass value
for i, mass in enumerate(masses):
    plt.text(i, mass, f'{quark_names[i]}\n{mass:.1f} MeV',
             ha='center', va='bottom', fontsize=10)

plt.legend()
plt.grid(True, which="both", ls="--", c='gray')
plt.show()
```

```python
"""
Numerical solver for Modular ADM Cosmology Friedmann equations.

This script will define the placeholder modular functions F(tau), U(tau), V(tau),
K_tautaubar(tau), and their derivatives, which are needed for the ODE system.
"""

import numpy as np

# --- Placeholder Modular Functions and their Derivatives ---

def F_func(tau, params):
    """Modular coupling function F(tau, taubar)."""
    # Example: F(tau, taubar) = F0 (constant)
    # More complex: F(tau, taubar) = 1.0 + params["c_F"] * (tau.imag - params["tau_I0"])**2
    return params.get("F0", 1.0)

def F_deriv_tau(tau, params):
    """Derivative of F with respect to tau."""
    # For F(tau, taubar) = F0, derivative is 0
    # If F = 1.0 + params["c_F"] * ( (tau - np.conjugate(tau))/(2j) - params["tau_I0"])**2
    # dF/dtau = params["c_F"] * 2 * ( (tau - np.conjugate(tau))/(2j) - params["tau_I0"]) * (1/(2j))
    # F_val = F_func(tau, params)
    # if F_val == params.get("F0", 1.0):
    #     return 0.0 + 0.0j
    # else:
    #     # Example for F = 1 + cF * (tau_I - tau_I0)^2
    #     # tau_I = (tau - np.conjugate(tau)) / (2j)
    #     # d(tau_I)/dtau = 1/(2j)
    #     # dF/dtau = cF * 2 * (tau_I - params["tau_I0"]) * (1/(2j))
    #     # return params["c_F"] * (tau.imag - params["tau_I0"]) * (-1j) # Simplified
    return 0.0 + 0.0j

def F_deriv_taubar(tau, params):
    """Derivative of F with respect to taubar."""
    # For F(tau, taubar) = F0, derivative is 0
    # If F = 1.0 + params["c_F"] * ( (tau - np.conjugate(tau))/(2j) - params["tau_I0"])**2
    # dF/dtaubar = params["c_F"] * 2 * ( (tau - np.conjugate(tau))/(2j) - params["tau_I0"]) * (-1/(2j))
    # F_val = F_func(tau, params)
    # if F_val == params.get("F0", 1.0):
    #     return 0.0 + 0.0j
    # else:
    #     # Example for F = 1 + cF * (tau_I - tau_I0)^2
    #     # d(tau_I)/dtaubar = -1/(2j)
    #     # dF/dtaubar = cF * 2 * (tau_I - params["tau_I0"]) * (-1/(2j))
    #     # return params["c_F"] * (tau.imag - params["tau_I0"]) * (1j) # Simplified
    return 0.0 + 0.0j

def U_func(tau, params):
    """Modular potential U(tau, taubar) (related to cosmological constant)."""
    # Example: U(tau, taubar) = U0 (constant)
    return params.get("U0", 0.01) # Example: Lambda_0 / (16*pi*G0)

def U_deriv_tau(tau, params):
    """Derivative of U with respect to tau."""
    # For U(tau, taubar) = U0, derivative is 0
    return 0.0 + 0.0j

def U_deriv_taubar(tau, params):
    """Derivative of U with respect to taubar."""
    # For U(tau, taubar) = U0, derivative is 0
    return 0.0 + 0.0j

def V_func(tau, params):
    """Modular potential V(tau, taubar) for the scalar field tau."""
    # Example: V(tau, taubar) = 0.5 * m_tau^2 * (tau_I - tau_I0)^2
    m_tau_sq = params.get("m_tau", 0.1)**2
    tau_I0 = params.get("tau_I0", 1.0)
    return 0.5 * m_tau_sq * (tau.imag - tau_I0)**2

def V_deriv_tau(tau, params):
    """Derivative of V with respect to tau."""
    # V = 0.5 * m_tau^2 * ( (tau - np.conjugate(tau))/(2j) - tau_I0 )**2
    # dV/dtau = 0.5 * m_tau^2 * 2 * ( (tau - np.conjugate(tau))/(2j) - tau_I0 ) * (1/(2j))
    #         = m_tau^2 * ( tau.imag - tau_I0 ) * (-0.5j)
    m_tau_sq = params.get("m_tau", 0.1)**2
    tau_I0 = params.get("tau_I0", 1.0)
    return m_tau_sq * (tau.imag - tau_I0) * (-0.5j)

def V_deriv_taubar(tau, params):
    """Derivative of V with respect to taubar."""
    # dV/dtaubar = 0.5 * m_tau^2 * 2 * ( (tau - np.conjugate(tau))/(2j) - tau_I0 ) * (-1/(2j))
    #            = m_tau^2 * ( tau.imag - tau_I0 ) * (0.5j)
    m_tau_sq = params.get("m_tau", 0.1)**2
    tau_I0 = params.get("tau_I0", 1.0)
    return m_tau_sq * (tau.imag - tau_I0) * (0.5j)

def Kttb_func(tau, params):
    """Metric on the target space of tau, K_tautaubar(tau, taubar)."""
    # Example: K_tautaubar = k0 / (Im(tau))^2
    k0 = params.get("k0", 1.0)
    if tau.imag == 0:
        # Avoid division by zero, though tau.imag should be > 0 for physical tau
        return k0 / (1e-10) # A very small number instead of zero
    return k0 / (tau.imag**2)

def Kttb_deriv_tau(tau, params):
    """Derivative of K_tautaubar with respect to tau."""
    # K = k0 / ( (tau - np.conjugate(tau))/(2j) )**2 = -4 * k0 / (tau - np.conjugate(tau))**2
    # dK/dtau = -4 * k0 * (-2) * (tau - np.conjugate(tau))**(-3) * 1
    #         = 8 * k0 / ( (2j * tau.imag) )**3 = 8 * k0 / (-8j * tau.imag**3)
    #         = k0 * 1j / (tau.imag**3)
    k0 = params.get("k0", 1.0)
    if tau.imag == 0:
        return 0.0 + 0.0j # Or handle error
    return k0 * (1j) / (tau.imag**3)

def Kttb_deriv_taubar(tau, params):
    """Derivative of K_tautaubar with respect to taubar."""
    # dK/dtaubar = -4 * k0 * (-2) * (tau - np.conjugate(tau))**(-3) * (-1)
    #            = -8 * k0 / ( (2j * tau.imag) )**3 = -8 * k0 / (-8j * tau.imag**3)
    #            = k0 * (-1j) / (tau.imag**3)
    k0 = params.get("k0", 1.0)
    if tau.imag == 0:
        return 0.0 + 0.0j # Or handle error
    return k0 * (-1j) / (tau.imag**3)

# --- Initial Conditions and Parameters ---

def get_initial_conditions_and_params():
    """Returns a dictionary of initial conditions and model parameters."""
    G0 = 1.0
    rho_m0 = 0.3
    rho_r0 = 5e-5
    a_initial = 1e-3
    tau_R_initial = 1.0
    tau_I_initial = 1.5
    tau_R_dot_initial = 0.0
    tau_I_dot_initial = 0.0
    initial_conditions = [a_initial, tau_R_initial, tau_I_initial, tau_R_dot_initial, tau_I_dot_initial]
    model_params = {
        "G0": G0, "rho_m0": rho_m0, "rho_r0": rho_r0,
        "F0": 1.0, "U0": 0.001, "m_tau": 0.01,
        "tau_I0": 1.5, "k0": 1.0, "c_F": 0.01
    }
    t_initial = 0.01
    t_final = 10.0
    num_points = 500
    t_span = np.linspace(t_initial, t_final, num_points)
    return initial_conditions, model_params, t_span

from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

# --- Coupled ODE System ---

def ode_system(t, y, params):
    a, tau_R, tau_I, tau_R_dot, tau_I_dot = y
    tau = tau_R + 1j * tau_I
    tau_dot = tau_R_dot + 1j * tau_I_dot
    G0 = params["G0"]
    rho_m0 = params["rho_m0"]
    rho_r0 = params["rho_r0"]
    eight_pi_G0 = 8 * np.pi * G0
    rho_m = rho_m0 / (a**3)
    rho_r = rho_r0 / (a**4)
    F = F_func(tau, params)
    F_tau = F_deriv_tau(tau, params)
    F_taubar = F_deriv_taubar(tau, params)
    U = U_func(tau, params)
    U_taubar = U_deriv_taubar(tau, params)
    V = V_func(tau, params)
    V_taubar = V_deriv_taubar(tau, params)
    K_ttb = Kttb_func(tau, params)
    K_ttb_tau = Kttb_deriv_tau(tau, params)
    K_ttb_taubar = Kttb_deriv_taubar(tau, params)
    term_H_coeff_B = 3 * np.real(F_tau * tau_dot)
    term_H_const_C = eight_pi_G0 * (rho_m + rho_r) + 0.5 * K_ttb * np.abs(tau_dot)**2 + V + U
    if np.abs(F) < 1e-10:
        print(f"Warning: F is close to zero ({F}) at t={t}, a={a}")
        H = 1e-10
    elif np.abs(term_H_coeff_B) < 1e-10:
        val_for_sqrt_H = term_H_const_C / (3 * F)
        if val_for_sqrt_H < 0:
            print(f"Warning: Negative value for H^2 ({val_for_sqrt_H}) at t={t}, a={a}. Setting H to small value.")
            H = 1e-10
        else:
            H = np.sqrt(val_for_sqrt_H)
    else:
        coeff_A_H_quad = 3 * F
        discriminant_H = term_H_coeff_B**2 + 4 * coeff_A_H_quad * term_H_const_C
        if discriminant_H < 0:
            print(f"Warning: Negative discriminant for H ({discriminant_H}) at t={t}, a={a}. Setting H to small value.")
            H = 1e-10
        else:
            H = (-term_H_coeff_B + np.sqrt(discriminant_H)) / (2 * coeff_A_H_quad)
    if H < 0 and a > 1e-6:
        print(f"Warning: Negative H ({H}) calculated at t={t}, a={a}. Assuming expansion and taking abs(H) or handling error.")
        H = np.abs(H)
    if np.abs(K_ttb) < 1e-10:
        print(f"Warning: K_ttb is close to zero ({K_ttb}) at t={t}, tau_I={tau.imag}")
        tau_ddot = 0.0 + 0.0j
    else:
        term1_tau_ddot = (K_ttb_tau * tau_dot + K_ttb_taubar * np.conjugate(tau_dot)) * tau_dot
        term2_tau_ddot = 3 * H * K_ttb * tau_dot
        term3_tau_ddot = V_taubar
        term4_tau_ddot = U_taubar / (eight_pi_G0)
        tau_ddot = - (1/K_ttb) * (term1_tau_ddot + term2_tau_ddot + term3_tau_ddot + term4_tau_ddot)
    a_dot = H * a
    tau_R_ddot = tau_ddot.real
    tau_I_ddot = tau_ddot.imag
    return [a_dot, tau_R_dot, tau_I_dot, tau_R_ddot, tau_I_ddot]

# --- Numerical Solver Implementation ---

def solve_modular_cosmology(initial_conditions, model_params, t_span):
    sol = solve_ivp(
        fun=ode_system,
        t_span=(t_span[0], t_span[-1]),
        y0=initial_conditions,
        t_eval=t_span,
        args=(model_params,),
        method='RK45',
        dense_output=True,
        rtol=1e-6,
        atol=1e-9
    )
    return sol

if __name__ == '__main__':
    # Test placeholder functions
    _test_params = {"F0":1.0,"U0":0.01,"m_tau":0.1,"tau_I0":1.5,"k0":1.0,"c_F":0.1}
    _test_tau = 1.0 + 2.0j
    print(f"Testing with tau = {_test_tau}, params = {_test_params}")
    print(f"F(tau) = {F_func(_test_tau, _test_params)}")
    print(f"V(tau) = {V_func(_test_tau, _test_params)}")
    print(f"K_ttb(tau) = {Kttb_func(_test_tau, _test_params)}")

    # Get initial conditions and parameters
    ic, mp, ts = get_initial_conditions_and_params()
    print("\n--- Initial Conditions & Parameters ---")
    print(f"Initial y_vector: {ic}")
    print(f"Model Parameters: {mp}")
    print(f"Time Span (start, end, num_points): {ts[0]}, {ts[-1]}, {len(ts)}")

    # Test the ODE system function with initial conditions
    print("\n--- Testing ode_system with initial conditions ---")
    dy_dt_initial = ode_system(ts[0], ic, mp)
    print(f"dy/dt at t_initial: {dy_dt_initial}")

    # Solve the ODE system
    print("\n--- Solving ODE system ---")
    solution = solve_modular_cosmology(ic, mp, ts)
    print(f"Solver status: {solution.status}")
    print(f"Solver message: {solution.message}")
    print(f"Number of time steps evaluated: {len(solution.t)}")
    print(f"Shape of solution y: {solution.y.shape}")

    # Extract solutions
    a_sol = solution.y[0]
    tau_R_sol = solution.y[1]
    tau_I_sol = solution.y[2]

    # Plotting
    plt.figure(figsize=(12, 8))
    plt.subplot(2, 2, 1)
    plt.plot(solution.t, a_sol)
    plt.xlabel("Time (t)")
    plt.ylabel("Scale Factor a(t)")
    plt.title("Scale Factor Evolution")
    plt.grid(True)

    plt.subplot(2, 2, 2)
    plt.plot(solution.t, tau_R_sol)
    plt.xlabel("Time (t)")
    plt.ylabel("Re(tau)")
    plt.title("Re(tau) Evolution")
    plt.grid(True)

    plt.subplot(2, 2, 3)
    plt.plot(solution.t, tau_I_sol)
    plt.xlabel("Time (t)")
    plt.ylabel("Im(tau)")
    plt.title("Im(tau) Evolution")
    plt.grid(True)

    plt.subplot(2, 2, 4)
    plt.plot(tau_R_sol, tau_I_sol)
    plt.xlabel("Re(tau)")
    plt.ylabel("Im(tau)")
    plt.title("Phase Space (tau_R vs tau_I)")
    plt.grid(True)
    plt.axis('equal')

    plt.tight_layout()
    # Changed save path for Colab compatibility
    plt.savefig("/content/modular_cosmology_solution_preview.png")
    print("\nSolution preview plot saved to /content/modular_cosmology_solution_preview.png")
    # In Colab, to display the plot directly after saving, you can use:
    # from IPython.display import Image
    # Image('/content/modular_cosmology_solution_preview.png')

```
