```python
import numpy as np

# General spiral mass formula
def spiral_mass(n, A, B, C, Omega, Phi):
    log_m = A + B * n + C * np.sin(Omega * n + Phi)
    return 10 ** log_m

#-------------------------
# Parameters (Refined/Best-Fit)
#-------------------------

# Charged leptons (Refined)
A_lep = -3.289668
B_lep = 1.999967
C_lep = 0.462226
Omega_lep = 3.883130
Phi_lep = 3.145729

# Up-type quarks (From earlier fit/documented values)
A_u = -2.549045
B_u = 2.433780
C_u = 0.219624
Omega_u = 4.266376
Phi_u = 3.658490

# Down-type quarks (From earlier fit/documented values)
A_d = -2.111075
B_d = 1.250780
C_d = 0.276398
Omega_d = 4.669596
Phi_d = 5.381334

# Neutrinos (From reparametrized spiral fit)
A_nu = -3.00132
B_nu = 0.96697
C_nu = 0.23948
Omega_nu = 5.06552
Phi_nu = 1.12018

#-------------------------
# Particle Naming and generations
#-------------------------

generations = [0, 1, 2]

lepton_names = ['e', 'mu', 'tau']
up_quark_names = ['u', 'c', 't']
down_quark_names = ['d', 's', 'b']
neutrino_names = ['nu1', 'nu2', 'nu3']

#-------------------------
# Predict mass function for particle family
#-------------------------

def predict_family(A, B, C, Omega, Phi, names, unit='GeV'):
    masses = []
    for n, name in zip(generations, names):
        m = spiral_mass(n, A, B, C, Omega, Phi)
        masses.append((name, m))
    return masses

#-------------------------
# Predict all
#-------------------------

masses_leptons = predict_family(A_lep, B_lep, C_lep, Omega_lep, Phi_lep, lepton_names)
masses_up_quarks = predict_family(A_u, B_u, C_u, Omega_u, Phi_u, up_quark_names)
masses_down_quarks = predict_family(A_d, B_d, C_d, Omega_d, Phi_d, down_quark_names)
masses_neutrinos = predict_family(A_nu, B_nu, C_nu, Omega_nu, Phi_nu, neutrino_names)

#-------------------------
# Print results with formatting
#-------------------------

def format_mass(m, particle_type):
    # Neutrino masses in eV, others in GeV with MeV fallback
    if particle_type == 'neutrino':
        return f"{m*1e9:.3f} meV"  # convert eV to meV
    else:
        if m < 0.001:
            return f"{m*1e3:.3f} MeV"
        else:
            return f"{m:.3f} GeV"

def print_masses(title, mass_list, particle_type='charged'):
    print(f"\n{title}:")
    for name, mass in mass_list:
        print(f"  {name:<4} : {format_mass(mass, particle_type)}")

# Output all masses
print_masses("Charged leptons masses", masses_leptons, 'charged')
print_masses("Up-type quark masses", masses_up_quarks, 'charged')
print_masses("Down-type quark masses", masses_down_quarks, 'charged')
print_masses("Neutrino masses", masses_neutrinos, 'neutrino')
```
