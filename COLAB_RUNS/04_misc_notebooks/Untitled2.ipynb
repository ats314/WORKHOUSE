```python
# Define extended n ranges for up-type and down-type quarks to visualize a possible 4th generation
n_up = np.linspace(0, 7, 1000)       # even n for up-type: u, c, t, t′ (0, 3, 5, 7)
n_down = np.linspace(0, 7, 1000)     # odd n for down-type: d, s, b, b′ (1, 2, 4, 6)

# Up-type spiral
theta_up = omega_q * n_up + phi_q
log_mass_up = lambda_q * n_up + epsilon_q * np.sin(theta_up) + log10_K_q
x_up = log_mass_up * np.cos(theta_up)
y_up = log_mass_up * np.sin(theta_up)
z_up = log_mass_up

# Down-type spiral (π phase offset)
theta_down = omega_q * n_down + phi_q + np.pi
log_mass_down = lambda_q * n_down + epsilon_q * np.sin(theta_down) + log10_K_q
x_down = log_mass_down * np.cos(theta_down)
y_down = log_mass_down * np.sin(theta_down)
z_down = log_mass_down + 0.1  # small z-offset for clarity

# Actual quark positions
up_quark_labels = ['u', 'c', 't']
down_quark_labels = ['d', 's', 'b']
n_up_quarks = [0, 3, 5]
n_down_quarks = [1, 2, 4]

theta_up_pts = omega_q * np.array(n_up_quarks) + phi_q
log_mass_up_pts = lambda_q * np.array(n_up_quarks) + epsilon_q * np.sin(theta_up_pts) + log10_K_q
x_up_pts = log_mass_up_pts * np.cos(theta_up_pts)
y_up_pts = log_mass_up_pts * np.sin(theta_up_pts)
z_up_pts = log_mass_up_pts

theta_down_pts = omega_q * np.array(n_down_quarks) + phi_q + np.pi
log_mass_down_pts = lambda_q * np.array(n_down_quarks) + epsilon_q * np.sin(theta_down_pts) + log10_K_q
x_down_pts = log_mass_down_pts * np.cos(theta_down_pts)
y_down_pts = log_mass_down_pts * np.sin(theta_down_pts)
z_down_pts = log_mass_down_pts + 0.1

# Plot the true double helix
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot up- and down-type spirals
ax.plot(x_up, y_up, z_up, color='blue', label='Up-type Helix (+2/3)')
ax.plot(x_down, y_down, z_down, color='red', label='Down-type Helix (−1/3)')

# Mark known quark positions
for i, label in enumerate(up_quark_labels):
    ax.scatter(x_up_pts[i], y_up_pts[i], z_up_pts[i], color='blue', s=80)
    ax.text(x_up_pts[i], y_up_pts[i], z_up_pts[i] + 0.05, label, color='blue')

for i, label in enumerate(down_quark_labels):
    ax.scatter(x_down_pts[i], y_down_pts[i], z_down_pts[i], color='red', s=80)
    ax.text(x_down_pts[i], y_down_pts[i], z_down_pts[i] + 0.05, label, color='red')

# Plot settings
ax.set_title("True Double Helix Spiral with Generational Rotations")
ax.set_xlabel("x = log(m)·cos(θ)")
ax.set_ylabel("y = log(m)·sin(θ)")
ax.set_zlabel("z = log(mass)")
ax.legend()
plt.tight_layout()
plt.show()
```
