```python
import numpy as np

# Locked HDTM parameters
lambda_param = 0.531
epsilon = 0.123
omega = 3.883
phi = -0.871
K = 0.0859

# Particle data: (name, n, Y, experimental mass in MeV)
particles = [
    ("u", 0, +1, 2.16),
    ("c", 1, +1, 1270),
    ("t", 2, +1, 172900),
    ("d", 0, -1, 4.67),
    ("s", 1, -1, 93.4),
    ("b", 2, -1, 4180),
    ("e", 0, 0, 0.511),
    ("mu", 1, 0, 105.7),
    ("tau", 2, 0, 1777)
]

# Compute predicted masses
print(f"{'Particle':>6} | {'Predicted (MeV)':>15} | {'Actual (MeV)':>15} | {'Error (%)':>10}")
print("-" * 55)

for name, n, Y, m_exp in particles:
    log_m = (
        lambda_param * n
        + epsilon * np.sin(omega * n + phi + Y * np.pi / 3)
        + np.log10(K * (2 ** Y))
    )
    m_pred = 10 ** log_m
    error = abs(m_pred - m_exp) / m_exp * 100
    print(f"{name:>6} | {m_pred:15.3f} | {m_exp:15.3f} | {error:10.2f}")
```
