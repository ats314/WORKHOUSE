```python
import sympy as sp
import numpy as np

# Define modular function scaling f(Y)
Y = sp.symbols('Y', real=True)  # Hypercharge label (-1, 0, +1)

# Define a specific form for omega_Y based on double-torus geometry
# This connects hypercharge directly to the geometry
omega_Y = sp.Function('omega_Y')(Y)
# A concrete implementation could be:
omega_Y = sp.pi/3 * sp.log(sp.Abs(Y**2 + sp.I))  # Example modular 1-form

# Modular function scaling (replaces 2^Y)
f_Y = sp.exp(sp.integrate(omega_Y, Y))

# Define SU(N) gauge group dimension
N = sp.symbols('N', integer=True, positive=True)

# Define lattice fields as proper matrix symbols
U = sp.MatrixSymbol('U', N, N)   # SU(N) gauge link variables
F = sp.MatrixSymbol('F', N, N)   # Field strength tensor
psi = sp.MatrixSymbol('psi', N, N)  # Fermion fields

# Lattice parameters
mu, nu = sp.symbols('mu nu', integer=True)
x = sp.symbols('x', integer=True)
L1, L2 = sp.symbols('L1 L2', integer=True, positive=True)  # Torus cycles
beta = sp.symbols('beta', positive=True)  # Inverse temperature

# Define the modular-invariant metric on the double-torus
# This encodes the genus-2 geometry with Y-dependent skewing
tau = sp.symbols('tau', complex=True)  # Modular parameter
G_mu_nu = sp.Matrix([[1, sp.re(tau)],
                     [sp.re(tau), sp.Abs(tau)**2]])

# Phase factor for twisted boundary conditions (encodes π/3 rotation)
phase_factor = sp.exp(sp.I * Y * sp.pi/3)

# Define the lattice action with modular function scaling
# Gauge kinetic term with genus-2 modulated metric and modular scaling
S_gauge = sp.Sum(
    sp.Trace(F * F) * G_mu_nu[mu, nu] * f_Y,
    (mu, 0, 1), (nu, 0, 1), (x, 0, L1*L2 - 1)
)

# Fermion kinetic term with modular scaling
S_fermion = sp.Sum(
    sp.Trace(psi * (U * psi * U.adjoint() - psi)) * f_Y,
    (mu, 0, 1), (x, 0, L1*L2 - 1)
)

# Full action
S_total = S_gauge + S_fermion

# Display the result
print("Modular function scaling f(Y):")
sp.pprint(f_Y)

print("\nTotal lattice action with modular scaling:")
sp.pprint(S_total)

# Test the action for specific Y values
for y_val in [-1, 0, 1]:
    print(f"\nEvaluating action components at Y = {y_val}:")
    f_y_val = f_Y.subs(Y, y_val)
    print(f"f({y_val}) = {float(f_y_val.evalf())}")

    # Compare with original 2^Y scaling
    original_scaling = 2**y_val
    print(f"Original 2^{y_val} = {original_scaling}")
    print(f"Ratio: {float(f_y_val.evalf())/original_scaling}")
```
