```python
import numpy as np

# Unified parameters
φ = (1 + np.sqrt(5))/2  # Golden ratio
ω = 2*np.pi/φ           # 3.883

def quark_mass(n, quark_type):
    if quark_type == 'up':
        λ = φ  # 1.618
        ε = 0.5
        phase = np.pi
        K = 10**-(φ**2)
    else:
        λ = 2/φ  # 1.236
        ε = 1/(2*φ)  # 0.309
        phase = np.pi + np.arccos(1/φ)
        K = 10**-φ

    # Optional dynamic coupling adjustment
    ε_adjusted = ε * (1 + 0.1*np.sin(np.pi*n/2))

    return 10**(λ*n + ε_adjusted*np.sin(ω*n + phase)) * K

# Calculate masses for all quarks
quark_indices = {
    'u': 1, 'd': 1,
    'c': 2, 's': 3,
    't': 6, 'b': 5
}

print("Quark Mass Predictions (GeV):")
print("-" * 50)
print("Quark | Index | Predicted Mass | Experimental Mass | % Diff")
print("-" * 50)

exp_masses = {
    'u': 0.0023, 'd': 0.0048,
    'c': 1.27, 's': 0.095,
    't': 173.0, 'b': 4.18
}

for q, idx in quark_indices.items():
    q_type = 'up' if q in ['u', 'c', 't'] else 'down'
    pred_mass = quark_mass(idx, q_type)
    exp_mass = exp_masses[q]
    diff_pct = 100 * (pred_mass - exp_mass) / exp_mass
    print(f"{q:5} | {idx:5} | {pred_mass:14.6f} | {exp_mass:16.6f} | {diff_pct:+6.2f}%")
```
