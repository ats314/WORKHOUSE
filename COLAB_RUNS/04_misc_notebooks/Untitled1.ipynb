```python
import matplotlib.pyplot as plt

# Canonical parameters
lambda_q = 0.9931
epsilon_q = 0.35
omega_q = 3.883
phi_q = 7.754687
logK_q = -3.040715

# Quark names and indices
strong_pairs = [("u", 0), ("d", 1), ("c", 2), ("s", 3), ("t", 4), ("b", 5)]
highlight_pairs = [("u", "d"), ("c", "s"), ("t", "b")]
quark_positions = {}

# Compute theta and log-mass for each quark
for name, n in strong_pairs:
    theta = omega_q * n + phi_q
    log_mass = lambda_q * n + epsilon_q * np.sin(theta) + logK_q
    x = log_mass * np.cos(theta)
    y = log_mass * np.sin(theta)
    quark_positions[name] = (x, y)

# Plotting
plt.figure(figsize=(8, 8))
for q1, q2 in highlight_pairs:
    x_vals = [quark_positions[q1][0], quark_positions[q2][0]]
    y_vals = [quark_positions[q1][1], quark_positions[q2][1]]
    plt.plot(x_vals, y_vals, 'r--', linewidth=1.5)

# Plot points
for name, (x, y) in quark_positions.items():
    plt.plot(x, y, 'o', label=name)
    plt.text(x + 0.01, y + 0.01, name, fontsize=12)

plt.title("Top-Down View of Quark Spiral — Strong CKM Pairs Only")
plt.xlabel("x (log mass · cosθ)")
plt.ylabel("y (log mass · sinθ)")
plt.axis('equal')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
