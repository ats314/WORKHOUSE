```python
import numpy as np

# Geodesic lengths
L = np.array([1.0, 1.5, 0.2])
# Target masses
m = np.array([0.002, 1.3, 173])

# Build the matrix
A = np.vstack([np.ones(3), L, L**2]).T

# Solve for coefficients
coeffs = np.linalg.solve(A, m)
c0, c1, c2 = coeffs

print(f"c0 = {c0:.6f}, c1 = {c1:.6f}, c2 = {c2:.6f}")
```
