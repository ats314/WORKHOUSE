```python
import numpy as np
import matplotlib.pyplot as plt

# Spiral parameters
a = 0.15  # growth rate
theta_max = 6 * np.pi
num_points = 1000
theta = np.linspace(0, theta_max, num_points)

# Spiral phase offsets
spirals = {
    "Up-type spiral (Y=+1)": (0, 'red'),
    "Lepton spiral (Y=0)": (np.pi / 3, 'blue'),
    "Down-type spiral (Y=-1)": (2 * np.pi / 3, 'green')
}

# Plot setup
fig, ax = plt.subplots(figsize=(8, 8))
for label, (phase, color) in spirals.items():
    r = np.exp(a * theta)
    x = r * np.cos(theta + phase)
    y = r * np.sin(theta + phase)
    ax.plot(x, y, color=color, linewidth=2, label=label)

ax.set_title("Spiral Resonant Coil – Phase-Shifted Spirals (π/3 Offset)", fontsize=14)
ax.set_aspect('equal')
ax.axis('off')
ax.legend()
plt.tight_layout()
plt.show()
```
