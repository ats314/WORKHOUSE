```python
import mpmath as mp

mp.mp.dps = 30
A = 1; B = 1.1
λ_target = mp.mpf('2.433780')

def log_ratio_C(C):
    def integrand(x):
        return 1/mp.sqrt(abs((x**2 - A**2)*(x**2 - B**2)*(x**2 - C**2)))
    Π1 = 2 * mp.quad(integrand, [A, B])
    Π2 = 2 * mp.quad(integrand, [B, C])
    return mp.log(Π2/Π1, 10)

# Scan C
for C_test in [3, 5, 10, 20, 50, 100]:
    print(f"C = {C_test}, log_ratio = {log_ratio_C(C_test)}")
```
