```python
import numpy as np
from itertools import product

# Physical constants (CODATA 2018)
alpha_inv = 137.035999084  # fine-structure constant inverse
mu = 1836.15267343         # proton/electron mass ratio
mn_mp = 1.001378419        # neutron/proton mass ratio
mmu_me = 206.7682830       # muon/electron mass ratio
mtau_me = 3477.23          # tau/electron mass ratio

# Mathematical constants
pi = np.pi
e = np.e
phi = (1 + np.sqrt(5)) / 2
zeta3 = 1.202056903159594
sqrt2 = np.sqrt(2)
sqrt3 = np.sqrt(3)
ln2 = np.log(2)
lnpi = np.log(np.pi)

# List of constants
phys_consts = {
    'alpha_inv': alpha_inv,
    'mu': mu,
    'mn_mp': mn_mp,
    'mmu_me': mmu_me,
    'mtau_me': mtau_me
}
math_consts = {
    'pi': pi,
    'e': e,
    'phi': phi,
    'zeta3': zeta3,
    'sqrt2': sqrt2,
    'sqrt3': sqrt3,
    'ln2': ln2,
    'lnpi': lnpi
}

# Search parameters
tolerance = 0.01  # relative error
powers = range(-5, 6)  # exponents to try
A_vals = [1, 2, 6, 12, 24, 120, 720]  # simple integer factors

print("Searching for simple relations...")

for pname, pval in phys_consts.items():
    for cname1, c1 in math_consts.items():
        for cname2, c2 in math_consts.items():
            for cname3, c3 in math_consts.items():
                for a, b, c in product(powers, repeat=3):
                    for A in A_vals:
                        # Avoid all-zero exponents
                        if a == b == c == 0:
                            continue
                        try:
                            val = A * (c1**a) * (c2**b) * (c3**c)
                            relerr = abs(val - pval) / pval
                            if relerr < tolerance:
                                print(f"{pname} ≈ {A}*{cname1}^{a}*{cname2}^{b}*{cname3}^{c} = {val:.6g} (rel. err. {relerr:.2e})")
                        except Exception:
                            continue
                        # Try exponentials
                        try:
                            val = A * np.exp(a * c1 + b * c2 + c * c3)
                            relerr = abs(val - pval) / pval
                            if relerr < tolerance:
                                print(f"{pname} ≈ {A}*exp({a}*{cname1} + {b}*{cname2} + {c}*{cname3}) = {val:.6g} (rel. err. {relerr:.2e})")
                        except Exception:
                            continue
                        # Try logs if possible
                        try:
                            if c1 > 0 and c2 > 0 and c3 > 0:
                                val = A * (np.log(c1)**a) * (np.log(c2)**b) * (np.log(c3)**c)
                                relerr = abs(val - pval) / pval
                                if relerr < tolerance:
                                    print(f"{pname} ≈ {A}*log({cname1})^{a}*log({cname2})^{b}*log({cname3})^{c} = {val:.6g} (rel. err. {relerr:.2e})")
                        except Exception:
                            continue
```

```python
import mpmath
mpmath.mp.dps = 20  # high precision

tau_i = 1j
tau_w = mpmath.exp(2 * mpmath.pi * 1j / 3)

# j-invariant
j_i = mpmath.j(tau_i)
j_w = mpmath.j(tau_w)

# Dedekind eta
eta_i = mpmath.eta(tau_i)
eta_w = mpmath.eta(tau_w)

# Eisenstein series (E4, E6)
def E4(tau):
    q = mpmath.exp(2 * mpmath.pi * 1j * tau)
    return 1 + 240 * sum([n**3 * q**n / (1 - q**n) for n in range(1, 50)])

def E6(tau):
    q = mpmath.exp(2 * mpmath.pi * 1j * tau)
    return 1 - 504 * sum([n**5 * q**n / (1 - q**n) for n in range(1, 50)])

E4_i = E4(tau_i)
E4_w = E4(tau_w)
E6_i = E6(tau_i)
E6_w = E6(tau_w)

print("j(i) =", j_i)
print("j(w) =", j_w)
print("eta(i) =", eta_i)
print("eta(w) =", eta_w)
print("ln(eta(i)) =", mpmath.ln(eta_i))
print("ln(eta(w)) =", mpmath.ln(eta_w))
print("E4(i) =", E4_i)
print("E4(w) =", E4_w)
print("E6(i) =", E6_i)
print("E6(w) =", E6_w)
```
